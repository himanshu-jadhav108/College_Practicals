## Case Study for Data Minimization Using Smart Threshold Alert by Providing Privacy-by-Design

**Aim:** To study and implement the concept of Data Minimization using a Smart Threshold Alert system based on the Privacy-by-Design (PbD) approach in an IoT environment.

### Step 1: Simulate Edge Readings & Apply Smart Threshold

```python
import random
import matplotlib.pyplot as plt

random.seed(7)
THRESHOLD = 33.0  # Celsius
readings = [round(random.gauss(30, 3), 1) for _ in range(30)]
sent = []

print("Reading | Value | Action")
for i, value in enumerate(readings, 1):
    if value > THRESHOLD:
        sent.append((i, value))
        print(f" {i:>3}   | {value:>5} | ALERT! Data transmitted to cloud")
    else:
        print(f" {i:>3}   | {value:>5} | Normal - kept on device (not sent)")

total = len(readings)
reduction = 100 * (total - len(sent)) / total

print("\n--- Summary ---")
print(f"Total readings      : {total}")
print(f"Transmitted (alerts): {len(sent)}")
print(f"Not transmitted     : {total - len(sent)}")
print(f"Data reduction      : {reduction:.1f} %")
```

### Step 2: Visualize Data Minimization Transmission

```python
plt.figure(figsize=(8, 4))
plt.plot(range(1, total + 1), readings, "o-", color="gray", label="Sensor reading")
plt.axhline(THRESHOLD, color="red", linestyle="--", label=f"Threshold = {THRESHOLD}°C")
plt.scatter([s[0] for s in sent], [s[1] for s in sent], color="red", zorder=3, label="Transmitted")
plt.xlabel("Reading no.")
plt.ylabel("Temperature (°C)")
plt.title("Smart Threshold Alert - Data Minimization")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

### Result & Conclusion
**Result:** The edge thresholding strategy achieved an 86.7% data reduction, communicating strictly critical events.

**Conclusion:** Data Minimization embodies Privacy-by-Design by cutting unnecessary network bandwidth, cloud storage overhead, and privacy leakage risks.
