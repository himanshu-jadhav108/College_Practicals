## Design and Implement a Password-Protected Smart Sensor System that Allows Access to Sensor Data Only After Successful Authentication, Demonstrating Basic IoT Security Concepts

**Aim:** To design and implement a password-protected smart sensor system that allows access to sensor data only after successful authentication using a basic security algorithm.

### Step 1: Security Architecture & SHA-256 Hashing

```python
import hashlib
import random

STORED_HASH = hashlib.sha256("IoT@123".encode()).hexdigest()
MAX_ATTEMPTS = 3

def read_sensor():
    return {
        "temperature": round(random.uniform(26, 31), 1),
        "humidity": round(random.uniform(50, 65), 1)
    }

def authenticate(password):
    pw_hash = hashlib.sha256(password.encode()).hexdigest()
    return pw_hash == STORED_HASH
```

### Step 2: Test Authentication Workflow

```python
print(f"Stored SHA-256 hash: {STORED_HASH}")
print("=== Smart Sensor Login ===")

passwords_to_try = ["wrong1", "admin", "IoT@123"]
attempts_left = MAX_ATTEMPTS
granted = False

for pw in passwords_to_try:
    print(f"Enter password: {pw}")
    if authenticate(pw):
        print("Access Granted")
        print("Sensor Data:", read_sensor())
        granted = True
        break
    else:
        attempts_left -= 1
        print(f"Access Denied ({attempts_left} attempts left)")

if not granted:
    print("Account locked: too many failed attempts")
```

### Result & Conclusion
**Result:** The system securely authorized valid credentials using SHA-256 digests and denied unauthorized attempts with lockout protection.

**Conclusion:** Cryptographic password hashing and attempt throttling prevent unauthorized telemetry exfiltration and protect edge IoT device integrity.
