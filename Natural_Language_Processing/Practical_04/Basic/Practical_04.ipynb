### Step 1: Import Libraries

```python
import pandas as pd
import numpy as np 
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from tabulate import tabulate
```

### Step 2: Sample Text Corpus

```python
corpus = [
    "the movie was good and the story was great",
    "the movie was bad and boring",
    "the story was amazing and the acting was good",
    "boring movie with a bad story",
    "great acting and great story"
]

for i, doc in enumerate(corpus, 1):
    print(f"Doc{i}: {doc}")
```

### Step 3: One-Hot Encoding

```python
# Build vocabulary of unique words
vocabulary = sorted(set(word for doc in corpus for word in doc.split()))
print("Vocabulary:", vocabulary)

# One-hot encode the first sentence
sample_sentence = corpus[0].split()
one_hot = pd.DataFrame(0, index=sample_sentence, columns=vocabulary)
for word in sample_sentence:
    one_hot.loc[word, word] = 1

print("\nOne-Hot Encoding for Doc1:")
print(tabulate(one_hot, headers='keys', tablefmt='fancy_grid'))
```

### Step 4: Bag of Words (Document-Term Matrix)

```python
count_vectorizer = CountVectorizer()
bow_matrix       = count_vectorizer.fit_transform(corpus)

bow_df = pd.DataFrame(
    bow_matrix.toarray(),
    columns = count_vectorizer.get_feature_names_out(),
    index   = [f"Doc{i+1}" for i in range(len(corpus))]
)

print("\nBag of Words Representation:")
print(tabulate(bow_df, headers='keys', tablefmt='fancy_grid'))
```

### Step 5: TF-IDF (Term Frequency-Inverse Document Frequency)

```python
tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix     = tfidf_vectorizer.fit_transform(corpus)

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray().round(3),
    columns = tfidf_vectorizer.get_feature_names_out(),
    index   = [f"Doc{i+1}" for i in range(len(corpus))]
)

print("\nTF-IDF Representation:")
print(tabulate(tfidf_df, headers='keys', tablefmt='fancy_grid'))
```

### Step 6: Sparsity Analysis

```python
total_values = bow_df.shape[0] * bow_df.shape[1]
zero_values  = (bow_df == 0).sum().sum()
sparsity     = (zero_values / total_values) * 100

print(f"Total Cells in matrix : {total_values}")
print(f"Zero Cells in matrix  : {zero_values}")
print(f"Sparsity              : {sparsity:.2f}%")
```

### Step 7: Comparison Summary

```python
# Helper calculations
total_cells_onehot = len(sample_sentence) * len(vocabulary)
zero_cells_onehot  = len(sample_sentence) * (len(vocabulary) - len(sample_sentence))
sparsity_onehot    = zero_cells_onehot / total_cells_onehot * 100

comparison_df = pd.DataFrame({
    "Representation"       : ["One-Hot Encoding", "Bag of Words", "TF-IDF"],
    "Total Cells"          : [total_cells_onehot, total_values, total_values],
    "Zero Cells"           : [zero_cells_onehot, zero_values, zero_values],
    "Sparsity (%)"         : [f"{sparsity_onehot:.2f}", f"{sparsity:.2f}", f"{sparsity:.2f}"],
    "Captures Frequency"   : ["No", "Yes", "Yes"],
    "Captures Importance"  : ["No", "No", "Yes"],
    "Sparse?"              : ["Yes", "Yes", "Yes"]
})

print("\nComparison of Representations:")
print(tabulate(comparison_df, headers='keys', tablefmt='fancy_grid'))
```

### Final Output 

```python
print("Experiment Completed Successfully")
print()
print("Vocabulary size:", len(vocabulary))
print("Matrix shape (BoW):", bow_df.shape)
print("Sparsity: {:.2f}%".format(sparsity))
```
