### Step 01: Import Libraries and Download Resources

```python
import nltk
import spacy
import pandas as pd

# NLTK Resources
nltk.download('punkt', quiet=True)
nltk.download('stopwords', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('averaged_perceptron_tagger', quiet=True)
nltk.download('averaged_perceptron_tagger_eng', quiet=True)

from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk import pos_tag

# Load spaCy English Model (Used for Lemmatization)
try:
    nlp = spacy.load("en_core_web_sm")
except OSError:
    from spacy.cli import download
    download("en_core_web_sm")
    nlp = spacy.load("en_core_web_sm")
    
print("Libraries Loaded and Resources Downloaded Successfully!")
```

### Step 02: Sample Input Text 

```python
text = """Natural Language Processing enables computers to understand human language.
She sat quietly on the river bank, watching the boats go by.
Later, she walked to the bank to deposit her monthly savings.
I saw the man with the telescope standing near the old library.
The studies conducted last year revealed surprising results."""

print(text)
```

### Step 03: Sentence Tokenization 

```python
sentences = sent_tokenize(text)
print("Number of Sentences: ", len(sentences))
for i, s in enumerate(sentences, 1):
    print(f"Sentence {i}: {s}")
```

### Step 04: Word Tokenization 

```python
words = word_tokenize(text)
print("Total Word Tokens: ", len(words))
# To check each word token, uncomment the following lines
# for i, word in enumerate(words, 1):
#     print(f"Word {i}: {word}")

print(words)
```

### Step 05: Stopword Removal 

```python
stop_words = set(stopwords.words("english"))

filtered_words = [w for w in words if w.isalpha() and w.lower() not in stop_words]

print("Tokens before Stopword Removal: ", len([w for w in words if w.isalpha()]))
print("Tokens after Stopword Removal: ", len(filtered_words))
print("Filtered Words: ", filtered_words)
```

### Step 06: Stemming Vs Lemmatization 

```python
stemmer = PorterStemmer()
doc = nlp(text)

lemma_dict = {token.text.lower(): token.lemma for token in doc if token.is_alpha}

comparison = []
for w in filtered_words:
    stem  = stemmer.stem(w.lower())
    lemma = lemma_dict.get(w.lower(), w.lower())
    comparison.append((w, stem, lemma))

df_compare = pd.DataFrame(comparison, columns=["Word", "Stemmed", "Lemmatized"])
df_compare = df_compare.drop_duplicates().reset_index(drop=True)
print(df_compare)
```

### Step 07: POS Tagging

```python
pos_tags = pos_tag(word_tokenize(sentences[3]))  # POS tagging for the fourth sentence
print("Sentence: ", sentences[3])
print()

for word, pos in pos_tags:
    print(f"{word:15s} -> {pos}")
```

### Step 08: Ambiguity Examples 

```python
# Lexical ambiguity: word "bank" has two different meanings in the text
print("Lexial Ambiguity Example: ")
print("Sentence 1: ", sentences[1])
print("Sentence 2: ", sentences[2])
print("-> 'bank' means river bank in first, financial bank in second\n")

# Syntactic ambiguity: prepositional phrase attachment
print("Syntactic Ambiguity Example: ")
print("Sentence: ", sentences[3])
print("-> 'with the telescope' can attach to 'saw' (I used it) or 'man' (he had it)")
```
