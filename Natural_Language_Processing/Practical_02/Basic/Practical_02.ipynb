### Step 1: Import Library

```python
import re
```

### Step 2: Sample Text with Emails, URLs and Hashtags

```python
text = """
Contact me at himanshujadhav40@gmail.com or himanshu.jadhav@college.edu.in
Visit https://himanshu-jadhav-portfolio.vercel.app/ or http://github.com/himanshu-jadhav108
Follow #NLP #MachineLearning #Python2026 #ArtificialIntelligence for updates
Also check www.openai.com
"""

print(text)
```

### Step 3: Extract Emails using Regex

```python
email_pattern = r'[\w.-]+@[\w.-]+\.\w+'
emails = re.findall(email_pattern, text)

print("Extracted Emails:")
print(emails)
```

### Step 4: Extract URLs using Regex

```python
url_pattern = r'(?:https?://|www\.)[\w./-]+'
urls = re.findall(url_pattern, text)

print("Extracted URLs:")
print(urls)
```

### Step 5: Extract Hashtags using Regex

```python
hashtag_pattern = r'#\w+'
hashtags = re.findall(hashtag_pattern, text)

print("Extracted Hashtags: ")
print(hashtags)
```

### Step 6: Design DFA for binary strings ending with '01'

```python
# States: q0 = start, q1 = seen a 0, q2 = accepting state (ends with 01)

transition_table = {
    'q0' : {'0': 'q1', '1': 'q0'},
    'q1' : {'0': 'q1', '1': 'q2'},
    'q2' : {'0': 'q1', '1': 'q0'}
}

acceptance_state = 'q2'

print("Transition Table: ")
for state, moves in transition_table.items():
    print(f"{state} -> {moves}")
```

### Step 7: Simulate the DFA on a given input string

```python
def run_dfa(binary_string):
    current_state = 'q0'
    for symbol in binary_string:
        current_state = transition_table[current_state][symbol]
    return current_state == acceptance_state

test_strings = ['1101', '100', '10101', '1110', '0001', '0101']

print("\nDFA Simulation Results: (Strings ending with '01' are accepted)")
for i, s in enumerate(test_strings):
    result = run_dfa(s)
    print(f"Input {i+1}: {s} -> {'Accepted' if result else 'Rejected'}")
```

### Final Output 

```python
print("Experiment Completed Successfully")
print("="*35)
print("Emails   :", emails)
print("URLs     :", urls)
print("Hashtags :", hashtags)
print("DFA Results:", {s: ("ACCEPTED" if run_dfa(s) else "REJECTED") for s in test_strings})
```
