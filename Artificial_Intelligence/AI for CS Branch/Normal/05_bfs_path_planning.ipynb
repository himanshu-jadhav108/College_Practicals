# Practical 5: Implement Breadth-First Search for a Planning Problem

## Aim
Implement BFS to find a valid path for a robot moving through a grid containing obstacles.

```python
# Cell 1: Grid Environment and Breadth-First Search Path Planner Implementation
from collections import deque

def bfs_path_planning(grid, start, goal):
    """
    Finds the shortest obstacle-free path for a robot using BFS.
    grid: 2D list of characters ('.' open, '#' obstacle).
    start: (row, col)
    goal: (row, col)
    """
    rows = len(grid)
    cols = len(grid[0])

    if grid[start[0]][start[1]] == '#' or grid[goal[0]][goal[1]] == '#':
        return None, 0

    queue = deque([(start, [start])])
    visited = {start}
    directions = [(-1, 0), (1, 0), (0, -1), (0, 1)]  # Up, Down, Left, Right

    while queue:
        current, path = queue.popleft()

        if current == goal:
            return path, len(visited)

        r, c = current
        for dr, dc in directions:
            nr, nc = r + dr, c + dc
            neighbor = (nr, nc)
            if 0 <= nr < rows and 0 <= nc < cols:
                if grid[nr][nc] != '#' and neighbor not in visited:
                    visited.add(neighbor)
                    queue.append((neighbor, path + [neighbor]))

    return None, len(visited)

def render_grid_view(grid, start, goal, path=None):
    """Visualizes grid layout with Start (S), Goal (G), Obstacles (#), and Path (*)."""
    path_set = set(path) if path else set()
    output_lines = []
    for r in range(len(grid)):
        row_chars = []
        for c in range(len(grid[0])):
            pos = (r, c)
            if pos == start:
                row_chars.append('S')
            elif pos == goal:
                row_chars.append('G')
            elif pos in path_set:
                row_chars.append('*')
            else:
                row_chars.append(grid[r][c])
        output_lines.append("  " + " ".join(row_chars))
    return "\n".join(output_lines)
```

```python
# Cell 2: Path Planning Execution, Grid Visualization, and Verification

# Scenario 1: Solvable 6x6 grid navigation
grid_solvable = [
    ['.', '.', '.', '.', '.', '.'],
    ['.', '#', '#', '.', '.', '.'],
    ['.', '.', '#', '.', '#', '.'],
    ['.', '.', '#', '.', '#', '.'],
    ['.', '#', '#', '.', '.', '.'],
    ['.', '.', '.', '.', '#', '.']
]
start_pos = (0, 0)
goal_pos = (5, 5)

print("=" * 65)
print("SCENARIO 1: SOLVABLE ROBOT PATH PLANNING")
print("=" * 65)
print("Initial Grid Layout (S: Start, G: Goal, #: Obstacle, .: Open):")
print(render_grid_view(grid_solvable, start_pos, goal_pos))
print("-" * 65)

path, visited_count = bfs_path_planning(grid_solvable, start_pos, goal_pos)

if path:
    print("Planned Path on Grid (*: Robot Path):")
    print(render_grid_view(grid_solvable, start_pos, goal_pos, path))
    print("-" * 65)
    print(f"Path Coordinates : {' -> '.join(str(p) for p in path)}")
    print(f"Path Length      : {len(path) - 1} steps")
    print(f"Cells Visited    : {visited_count}")

    # Verification assertions
    assert path[0] == start_pos, "Path does not begin at Start position!"
    assert path[-1] == goal_pos, "Path does not terminate at Goal position!"
    assert all(grid_solvable[r][c] != '#' for r, c in path), "Path illegally traverses an obstacle!"
    print("Verification     : Path successfully reaches goal without crossing obstacles.")

# Scenario 2: Unsolvable / Completely blocked grid
grid_blocked = [
    ['.', '.', '.', '#', '.', '.'],
    ['.', '#', '.', '#', '.', '.'],
    ['.', '#', '.', '#', '#', '#'],
    ['.', '#', '.', '.', '.', '#'],
    ['.', '#', '#', '#', '#', '#'],
    ['.', '.', '.', '.', '#', '.']
]
print("\n" + "=" * 65)
print("SCENARIO 2: BLOCKED GRID (NO PATH AVAILABLE)")
print("=" * 65)
path_blocked, visited_blocked = bfs_path_planning(grid_blocked, start_pos, goal_pos)
print(render_grid_view(grid_blocked, start_pos, goal_pos))
print("-" * 65)
print(f"Search Result    : Path Found = {path_blocked is not None}")
print(f"Cells Visited    : {visited_blocked}")
assert path_blocked is None, "Verification Failed: Blocked grid should not return a path!"
print("Verification     : Correctly identified and reported that no path exists.")
print("Verification Status: ALL CHECKS PASSED")
```
