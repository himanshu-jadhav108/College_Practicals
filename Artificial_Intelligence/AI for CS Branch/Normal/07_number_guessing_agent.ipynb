# Practical 7: Design an Intelligent Agent to Guess a Number Using Higher/Lower Feedback

## Aim
Implement a number-guessing agent that uses feedback to identify a secret number efficiently.

```python
# Cell 1: Environment and Binary-Search Intelligent Agent Implementation

class NumberGuessingEnvironment:
    """Simulated environment holding a secret target number."""
    def __init__(self, secret_number, low=1, high=100):
        self.secret_number = secret_number
        self.low = low
        self.high = high

    def evaluate_guess(self, guess):
        """Returns Higher, Lower, or Correct feedback."""
        if guess == self.secret_number:
            return 'Correct'
        elif self.secret_number > guess:
            return 'Higher'
        else:
            return 'Lower'

class BinarySearchAgent:
    """Intelligent agent employing binary search strategy on interval [low, high]."""
    def __init__(self, low=1, high=100):
        self.low = low
        self.high = high

    def search(self, environment):
        history = []
        low, high = self.low, self.high
        guess_count = 0

        while low <= high:
            guess = (low + high) // 2
            guess_count += 1
            feedback = environment.evaluate_guess(guess)
            history.append({
                'step': guess_count,
                'interval': (low, high),
                'guess': guess,
                'feedback': feedback
            })

            if feedback == 'Correct':
                break
            elif feedback == 'Higher':
                low = guess + 1
            elif feedback == 'Lower':
                high = guess - 1

        return guess, guess_count, history
```

```python
# Cell 2: Automated Evaluation across Multiple Target Values and Boundary Verification

def evaluate_agent_for_target(secret_target, low=1, high=100):
    env = NumberGuessingEnvironment(secret_target, low, high)
    agent = BinarySearchAgent(low, high)
    found_number, total_guesses, trace = agent.search(env)

    print("=" * 65)
    print(f"SECRET TARGET: {secret_target} (Permitted Range: [{low}, {high}])")
    print("=" * 65)
    print(f"{'Step':<6} | {'Current Interval':<18} | {'Agent Guess':<12} | {'Feedback':<12}")
    print("-" * 65)
    for record in trace:
        print(f"{record['step']:<6} | {str(record['interval']):<18} | {record['guess']:<12} | {record['feedback']:<12}")
    print("-" * 65)
    print(f"Result: Secret {secret_target} identified in {total_guesses} guesses (<= 7 upper bound).")
    
    # Verification assertions
    assert found_number == secret_target, f"Verification Failed: {found_number} != {secret_target}"
    assert total_guesses <= 7, f"Verification Failed: Guesses exceeded theoretical bound ({total_guesses} > 7)"
    print("Verification: PASSED\n")

# Multiple test scenarios including boundary and intermediate values
evaluate_agent_for_target(secret_target=1)    # Lower boundary
evaluate_agent_for_target(secret_target=100)  # Upper boundary
evaluate_agent_for_target(secret_target=50)   # Midpoint
evaluate_agent_for_target(secret_target=42)   # Arbitrary test
evaluate_agent_for_target(secret_target=73)   # Arbitrary test
```
