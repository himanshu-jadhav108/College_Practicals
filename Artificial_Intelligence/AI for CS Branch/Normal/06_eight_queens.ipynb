# Practical 6: Implement Backtracking to Solve the 8-Queens Constraint Satisfaction Problem

## Aim
Solve the 8-Queens problem using backtracking while satisfying all queen-placement constraints.

```python
# Cell 1: 8-Queens Backtracking CSP Solver Implementation

def is_placement_safe(queens, row, col):
    """
    Checks if placing a queen at (row, col) conflicts with already placed queens.
    queens: list where queens[r] is the column index of queen placed in row r.
    """
    for prev_row in range(row):
        prev_col = queens[prev_row]
        # Check column conflict
        if prev_col == col:
            return False
        # Check diagonal conflicts
        if abs(prev_row - row) == abs(prev_col - col):
            return False
    return True

def solve_8_queens():
    """Solves the 8-Queens CSP using depth-first backtracking."""
    queens = [-1] * 8
    metrics = {'attempts': 0, 'backtracks': 0}

    def backtrack(row):
        if row == 8:
            return True

        for col in range(8):
            metrics['attempts'] += 1
            if is_placement_safe(queens, row, col):
                queens[row] = col
                if backtrack(row + 1):
                    return True
                # Backtrack: revert placement
                queens[row] = -1
                metrics['backtracks'] += 1

        return False

    backtrack(0)
    return queens, metrics
```

```python
# Cell 2: Chessboard Visualization, Search Metrics, and Constraint Verification

def render_chessboard(queens):
    """Renders 8x8 chessboard with 'Q' for queens and '.' for empty squares."""
    lines = []
    lines.append("  +---+---+---+---+---+---+---+---+")
    for r in range(8):
        row_str = []
        for c in range(8):
            if queens[r] == c:
                row_str.append(" Q ")
            else:
                row_str.append(" . ")
        lines.append(f"  |{'|'.join(row_str)}| Row {r}")
        lines.append("  +---+---+---+---+---+---+---+---+")
    lines.append("    0   1   2   3   4   5   6   7  (Cols)")
    return "\n".join(lines)

def verify_constraints(queens):
    """Rigorous verification of 8-Queens constraint satisfaction."""
    if len(queens) != 8 or any(c == -1 for c in queens):
        return False, "Board does not contain 8 queens."
    if len(set(queens)) != 8:
        return False, "Column constraint violated: Multiple queens share a column."
    for r1 in range(8):
        for r2 in range(r1 + 1, 8):
            c1, c2 = queens[r1], queens[r2]
            if abs(r1 - r2) == abs(c1 - c2):
                return False, f"Diagonal constraint violated between ({r1},{c1}) and ({r2},{c2})."
    return True, "All placement constraints satisfied perfectly."

# Solve 8-Queens
solution_queens, search_metrics = solve_8_queens()

print("=" * 65)
print("8-QUEENS CONSTRAINT SATISFACTION PROBLEM (CSP) SOLUTION")
print("=" * 65)
print("Final 8x8 Chessboard Configuration:")
print(render_chessboard(solution_queens))
print("-" * 65)
queen_coords = [(r, solution_queens[r]) for r in range(8)]
print(f"Queen Coordinates (Row, Col): {queen_coords}")
print(f"Placement Attempts           : {search_metrics['attempts']}")
print(f"Backtracking Operations      : {search_metrics['backtracks']}")
print("-" * 65)

is_satisfied, check_msg = verify_constraints(solution_queens)
print(f"Constraint Check Result      : {check_msg}")
assert is_satisfied, f"Verification Failed: {check_msg}"
print("Verification Status          : ALL CHECKS PASSED")
```
