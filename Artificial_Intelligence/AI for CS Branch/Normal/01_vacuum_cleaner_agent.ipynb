# Practical 1: Design and Implement a Simple Reflex Agent for the Vacuum Cleaner World and Analyze Rational Behavior

## Aim
Implement a simple reflex vacuum-cleaner agent and demonstrate its behavior in a defined environment.

```python
# Cell 1: Environment Model and Simple Reflex Agent Implementation

class VacuumEnvironment:
    """
    Two-room environment (Rooms A and B) with discrete Clean/Dirty states.
    """
    def __init__(self, initial_location='A', initial_states=None):
        self.locations = ['A', 'B']
        self.agent_location = initial_location
        if initial_states is None:
            self.states = {'A': 'Dirty', 'B': 'Dirty'}
        else:
            self.states = dict(initial_states)

    def get_percept(self):
        """Returns the current location and cleanliness condition of that room."""
        return (self.agent_location, self.states[self.agent_location])

    def execute_action(self, action):
        """Applies the agent's action to update the environment state."""
        if action == 'Clean':
            self.states[self.agent_location] = 'Clean'
        elif action == 'Right':
            self.agent_location = 'B'
        elif action == 'Left':
            self.agent_location = 'A'
        elif action == 'NoOp':
            pass


class SimpleReflexAgent:
    """
    Simple reflex agent operating on condition-action rules.
    """
    def __init__(self):
        pass

    def decide_action(self, percept, environment_states):
        location, status = percept
        # Condition-Action Rules
        if status == 'Dirty':
            return 'Clean'
        elif location == 'A' and environment_states.get('B') == 'Dirty':
            return 'Right'
        elif location == 'B' and environment_states.get('A') == 'Dirty':
            return 'Left'
        else:
            return 'NoOp'
```

```python
# Cell 2: Simulation Execution and Behavioral Verification

def run_vacuum_simulation(scenario_name, initial_location, initial_states):
    print("=" * 65)
    print(f"Scenario: {scenario_name}")
    print("=" * 65)
    env = VacuumEnvironment(initial_location, initial_states)
    agent = SimpleReflexAgent()

    print(f"Initial State  : Agent at {env.agent_location} | Room A: {env.states['A']} | Room B: {env.states['B']}")
    print("-" * 65)

    step = 1
    action_history = []
    max_steps = 10

    while step <= max_steps:
        percept = env.get_percept()
        action = agent.decide_action(percept, env.states)
        action_history.append(action)

        print(f"Step {step:02d} | Location: {percept[0]} | Condition: {percept[1]:<5} -> Action: {action}")
        env.execute_action(action)
        print(f"        Updated State -> Room A: {env.states['A']} | Room B: {env.states['B']} | Agent at: {env.agent_location}")

        if action == 'NoOp':
            print("Agent decided NoOp (All rooms clean, rational objective accomplished).")
            break
        step += 1

    print("-" * 65)
    print(f"Final State    : Agent at {env.agent_location} | Room A: {env.states['A']} | Room B: {env.states['B']}")
    is_all_clean = all(status == 'Clean' for status in env.states.values())
    print(f"Verification   : All rooms clean: {is_all_clean} | Action sequence: {action_history}")
    assert is_all_clean, "Verification Failed: Environment not completely cleaned!"
    print("Behavior Status: RATIONAL & SUCCESSFUL\n")


# Scenario 1: Both rooms initially dirty, agent starts in Room A
run_vacuum_simulation("Both Rooms Dirty (Start at A)", 'A', {'A': 'Dirty', 'B': 'Dirty'})

# Scenario 2: Room A clean, Room B dirty, agent starts at A
run_vacuum_simulation("Room A Clean, Room B Dirty (Start at A)", 'A', {'A': 'Clean', 'B': 'Dirty'})

# Scenario 3: Both rooms already clean, agent starts at B
run_vacuum_simulation("Both Rooms Already Clean (Start at B)", 'B', {'A': 'Clean', 'B': 'Clean'})
```
